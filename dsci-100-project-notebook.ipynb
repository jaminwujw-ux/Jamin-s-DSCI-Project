This plot shows the proportion of teams that lost their first game at each seed. Low seeds (1 to 4) rarely lose, while high seeds (12 to 16) lose most of the time. So seed is very related to first round win/loss and should be a useful predictor in the KNN model.

```r
vis1 <- tidy_data |>
    ggplot(aes(x = seed, fill = first_round_loss)) +
    geom_bar(position = "fill") +
    labs(x = "Tournament Seed",
         y = "Proportion of Teams",
         fill = "Lost First Game?") +
    ggtitle("Proportion of Teams Losing Their First Game by Seed") +
    theme(text = element_text(size = 14))

vis1
```

This scatterplot shows both predictors together, with colour showing the outcome. Teams that lost their first game cluster at high seeds, while winners cluster at low seeds, with some overlap. Because the groups form a pattern in these two variables, nearest neighbours should be able to predict the outcome reasonably well.

```r
vis2 <- tidy_data |>
    ggplot(aes(x = reg_wins_pct)) +
    geom_histogram(binwidth = 5) +
    facet_wrap(~ first_round_loss,
               labeller = as_labeller(c("FALSE" = "Won First Game",
                                        "TRUE" = "Lost First Game"))) +
    labs(x = "Regular Season Win Percentage (%)",
         y = "Number of Teams") +
    ggtitle("Regular Season Win Percentage by First Game Result") +
    theme(text = element_text(size = 14))

vis2
```

This plot compares the regular season win percentage of teams that won versus lost their first game. Teams with win percentages above about 85% were much more likely to win, while teams below about 65% were more likely to lose. In the middle range (about 70 to 85%), the two groups look very similar, so overall win percentage is useful for predicting the outcome at extremities, but is likely less useful for mid-range teams.
